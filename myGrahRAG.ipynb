# 使用microsoft GraphRAG创建知识图谱

## INdex 索引
1. 将语料库接分为TextUnit，作为基础分析单元
2. 从 TextUnit 中提取所有实体、关系和关键主张
3. 图谱构建： 将提取出的信息转换成一张复杂的网——知识图谱
4. 社区检测（Community Detection）： 这是微软引入的高级技术。它利用算法（如 Leiden 算法）将紧密相关的实体聚集在一起形成“社区”。
5. 层次化总结： LLM 会为每个社区生成摘要，从底层细节到高层概述，形成一个分层的索引体系。

## 查询（检索与生成）
1. 全局搜索（Global Search）： 针对汇总性问题。它会扫描预先生成的“社区摘要”，从而给出跨文档、高层级的全面回答
2. 局部搜索（Local Search）： 针对特定实体的问题。它会结合知识图谱中的特定节点、其邻居节点以及原始文本块，进行深度的关联检索。

### 接入SEU算力平台

```python
from dotenv import load_dotenv
load_dotenv()  # 自动寻找 .env 文件并加载到 os.environ
```

```python
import os
from openai import OpenAI

try:
    client = OpenAI(
        api_key=os.getenv("DS_API_KEY"),
        base_url="https://api.deepseek.com",
    )

    completion = client.chat.completions.create(
        model="deepseek-chat",  
        messages=[
            {'role': 'user', 'content': '你是谁？你是哪款模型？'},

        ]
    )
    print(completion.choices[0].message.content)
except Exception as e:
    print(f"错误信息：{e}")
```
